
# Notebook 4 - Gráficos e roteiro para a banca

Este notebook concentra os gráficos que você pode usar na apresentação.


```python

import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
BASE = Path("..").resolve()
df = pd.read_csv(BASE / "data/processed/base_integrada.csv")
```

## Gráfico 1 - Produtos mais contratados

```python

produtos = (df[["id_contrato","produto_nome"]].drop_duplicates()
            .groupby("produto_nome")["id_contrato"].count()
            .sort_values(ascending=False).head(10))
plt.figure(figsize=(10,5))
produtos.plot(kind="bar")
plt.title("Top 10 produtos mais contratados")
plt.tight_layout()
plt.show()
```

## Gráfico 2 - Clientes com mais transações

```python

clientes = df.groupby("cliente_nome")["id_transacao"].count().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
clientes.plot(kind="bar")
plt.title("Top 10 clientes com mais transações")
plt.tight_layout()
plt.show()
```

## Gráfico 3 - Evolução mensal

```python

mensal = df.groupby("ano_mes")["transacao_valor"].sum().sort_index()
plt.figure(figsize=(12,5))
mensal.plot(kind="line")
plt.title("Evolução mensal do valor transacionado")
plt.tight_layout()
plt.show()
```
